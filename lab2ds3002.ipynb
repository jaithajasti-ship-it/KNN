```python
#QUESTION 3:

#reshaping single image

index = 0
print(X_test[index].shape)
X_test_vector = X_test[index].reshape(1, 784)
print(X_test_vector.shape)



#reshape all training data

import numpy as np

# Assume X_train has shape (num_train_samples, 28, 28)
num_train_samples = X_train.shape[0]

# Create an empty array to store flattened vectors
X_train_flat = np.zeros((num_train_samples, 28*28))

for i in range(num_train_samples):
    X_train_flat[i] = X_train[i].reshape(784)

print(X_train_flat.shape)



#scikit-learn classifier

# Flatten training and testing images
X_train_flat = X_train.reshape(X_train.shape[0], 28*28)
X_test_flat  = X_test.reshape(X_test.shape[0], 28*28)

print(X_train_flat.shape)
print(X_test_flat.shape)



#train and predict
from sklearn.neighbors import KNeighborsClassifier

# Initialize KNN classifier
knn = KNeighborsClassifier(n_neighbors=3)

# Train on flattened training data
knn.fit(X_train_flat, y_train)

# Predict on flattened test data
preds = knn.predict(X_test_flat)

print(preds[:10])  # Print first 10 predictions
```

```python
#QUESTION 4:

# Flatten images for scikit-learn
X_train_flat = X_train.reshape(X_train.shape[0], 28*28)
X_test_flat  = X_test.reshape(X_test.shape[0], 28*28)

print(X_train_flat.shape)  # (60000, 784)
print(X_test_flat.shape)   # (10000, 784)

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

neighbors_list = [1, 3, 5, 7, 9]
accuracies = []

for k in neighbors_list:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_flat, y_train)         # Train on training data
    preds = knn.predict(X_test_flat)       # Predict on test data
    acc = accuracy_score(y_test, preds)    # Compute accuracy
    accuracies.append(acc)
    print(f"k = {k}, Accuracy = {acc:.4f}")


import matplotlib.pyplot as plt

plt.plot(neighbors_list, accuracies, marker='o')
plt.xlabel("Number of neighbors (k)")
plt.ylabel("Accuracy on test set")
plt.title("kNN Accuracy vs k")
plt.show()
```

The model achieves its highest accuracy when k=3 with an accuracy of 0.9705 (97.1%). Therefore, the optimal choice for the number of neighbors is k=3.

```python
#QUESTION 5

#training kNN with optimal k

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Optimal number of neighbors
k_optimal = 3

# Initialize and train kNN
knn = KNeighborsClassifier(n_neighbors=k_optimal)
knn.fit(X_train_flat, y_train)

# Predict on test set
y_pred = knn.predict(X_test_flat)

#computing accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy on test set: {accuracy:.4f}")

# Compute confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Display it properly
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=range(10))
disp.plot(cmap=plt.cm.Blues)
plt.title(f"kNN Confusion Matrix (k={k_optimal})")
plt.show()
```

The model’s accuracy is 97.05%, which means it correctly classifies about 97 out of every 100 handwritten digits. Most of the predictions are correct, as shown by the strong diagonal in the confusion matrix. The few mistakes usually happen between digits that look similar.
Using k=3 works well because it balances underfitting and overfitting, and the remaining errors are mostly due to messy handwriting or noise in the raw pixel data.

In the RGB color model, each pixel is represented by three numbers between 0 and 255,
showing the amounts of red, green, and blue. Because of this, a color image can be stored as a 3D array with shape (H, W, 3), where H is height, W is width, and 3 is the number of color channels.
For example, a 32 x 32 color image would have shape (32, 32, 3). A single pixel might look like [128, 198, 56], which represents one specific color.
To use this image in a machine learning model, we flatten the 3D array into a single vector. Since 32 x 32 x 3 = 3072, each image becomes a row with 3072 values. If we have N images, the dataset will have shape (N, 3072).
After flattening, each row represents one image and each column represents one pixel value. This lets us treat images like regular tabular data and apply models such as kNN to classify or analyze them.
